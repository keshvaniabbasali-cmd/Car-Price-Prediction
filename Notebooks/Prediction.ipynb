```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import pickle
```

```python
with open("Notebooks/car_price_model.pkl", "rb") as file:
    model = pickle.load(file)

with open("Notebooks/Label_Encoder.pkl", "rb") as file:
    LabelEncoder = pickle.load(file)

with open("Notebooks/Scaler_and_OneHotEncoder.pkl", "rb") as file:
    Scaler_and_OneHotEncoder = pickle.load(file)
```

```python
input_data = {
    "year" : 2018,
    "manufacturer" : "bmw",
    "condition" : "fair",
    "fuel" : "diesel",
    "odometer" : 150000,
    "title_status" : "clean",
    "transmission" : "manual",
    "drive" : "fwd",
    "type" : "sedan",
    "paint_color" : "white",
    "car_age" : 8
}
```

```python
input_data["condition"] = LabelEncoder.transform([input_data["condition"]])
```

```python
input_data = Scaler_and_OneHotEncoder.transform(pd.DataFrame(input_data))
```

```python
model.predict(input_data)
```
